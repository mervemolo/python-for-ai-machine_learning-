```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    'Musteri_ID': range(1, 11),
    'Maas': [35000, 42000, 39000, 45000, 40000, 48000, 52000, 600000, 2000, 41000] # 600.000 ve 2.000 aykırı değer
})
```

```python
q1=df["Maas"].quantile(0.25)
q3=df["Maas"].quantile(0.75)
iqr=q3-q1
```

```python
alt_sinir=q1-(1.5*iqr)
ust_sinir=q3+(1.5*iqr)
```

```python
df_temiz=df[(df["Maas"]>=alt_sinir)&(df["Maas"]<=ust_sinir)]
```
