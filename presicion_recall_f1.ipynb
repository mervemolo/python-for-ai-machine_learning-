```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv("student_lifestyle_100k.csv")
```

```python
df.head(3)
```

```python
df=df.drop("Student_ID",axis=1)
```

```python
df=pd.get_dummies(df,columns=["Gender","Department"])
df
```

```python
y=df["Depression"]
x=df.drop("Depression",axis=1)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,train_size=0.9)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf=RandomForestClassifier()
```

```python
model=rf.fit(x_train,y_train)
model.score(x_test,y_test)
```

```python
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test,y_pred)
```

```python
dfs=pd.DataFrame()
dfs["sonuc"]=y_pred
dfs["sonuc"].value_counts()
dfs["gercek"]=y_test.values
```

```python
dfs["TrueBil"]=(dfs["gercek"]==True)& (dfs["sonuc"]==True)
```
