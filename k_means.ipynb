```python
import pandas as pd
from sklearn.cluster import KMeans
```

```python
data = {
    'Yas': [18, 22, 25, 45, 52, 58, 23, 49],
    'Harcama_Skoru': [80, 90, 85, 20, 15, 30, 95, 25] # 1-100 arası skor
}
df=pd.DataFrame(data)
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
```

```python
df["Küme"]=kmeans.fit_predict(df)
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
inertia_degerleri=[]
k_araligi=range(1,9)
```

```python
for k in k_araligi:
    kmeans=KMeans(n_clusters=k,random_state=42)
    kmeans.fit(df[['Yas', 'Harcama_Skoru']])
    inertia_degerleri.append(kmeans.inertia_)
```

```python
for k,inertia in zip(k_araligi,inertia_degerleri):
    print(f"K = {k} için Küme İçi Hata (Inertia): {inertia:.2f}")
```

```python
plt.plot(k_araligi, inertia_degerleri, marker='o')
plt.xlabel('Küme Sayısı (K)')
plt.ylabel('Küme İçi Hata (Inertia)')
plt.title('Dirsek Yöntemi (Elbow Method)')
plt.grid(True)
plt.show()
```
