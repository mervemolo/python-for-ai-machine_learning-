```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("hospital_readmission_risk_10000.csv")
```

```python
df.head(3)
```

```python
df=df.drop("patient_id",axis=1)
```

```python
df.head(3)
df=df.dropna()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
```

```python
df["readmission_label"]=le.fit_transform(df["readmission_risk"])
```

```python
y=df["readmission_label"]
x=df.drop(columns=["readmission_label","readmission_risk"])
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,train_size=0.79)
```

```python
from catboost import CatBoostClassifier
```

```python
cat_features=[]
for i in x.columns:
    if x[i].dtype=="str":
        cat_features.append(i)
```

```python
cb=CatBoostClassifier(
    learning_rate=0.5,
    loss_function="MultiClass",
    n_estimators=500
)


model=cb.fit(x_train,y_train,cat_features=cat_features)
```

```python
model.score(x_test,y_test)
```

```python
model.predict([[24,"Female",82,178,25.88,8,"COPD",8,10.4,146,0.55,'Emergency', 15, 0, 'Never', 'High', 'Low', 'Uninsured', 'Poor',
        'Strong', 'No']])
```
