```python
import pandas as pd
from sklearn.preprocessing import StandardScaler,MinMaxScaler
```

```python
df = pd.DataFrame({
    'Yas': [22, 45, 30, 50],
    'Gelir': [15000, 90000, 45000, 120000]
})
```

```python
sclaer_std=StandardScaler()
```

```python
df["Gelir_Standard"]=sclaer_std.fit_transform(df[["Gelir"]])
```

```python
scaler_minmax=MinMaxScaler()
df["Gelir_MinMax"]=scaler_minmax.fit_transform(df[["Gelir"]])
```

```python
from sklearn.impute import SimpleImputer
import numpy as np
import pandas as pd
```

```python
data = {
    'Yas': [22, 38, np.nan, 26, 35],           # 1 adet boş
    'Maas': [20000, 50000, 30000, np.nan, 45000],# 1 adet boş
    'Departman': ['IT', 'IK', 'IT', 'IK', np.nan] # 1 adet boş kategorik
}
```

```python
df=pd.DataFrame(data)
```

```python
num_imputer=SimpleImputer(strategy="median")
df[["Yas","Maas"]]=num_imputer.fit_transform(df[["Yas","Maas"]])
cat_imputer=SimpleImputer(strategy="most_frequent")
df[["Departman"]]=cat_imputer.fit_transform(df[["Departman"]])
```
