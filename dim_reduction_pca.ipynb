```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
X = pd.DataFrame(iris.data, columns=iris.feature_names)
```

```python
scaler=StandardScaler()
x_scaled=scaler.fit_transform(X)
```

```python
pca=PCA(n_components=2)
x_pca=pca.fit_transform(x_scaled)
```

```python
df_pca=pd.DataFrame(data=x_pca,columns=["PC1","PC2"])
```

```python
print("Orjinal veri boyutu:",X.shape)
print("PCA sonrası boyut:",df_pca.shape)
```

```python
varyans_oranlari=pca.explained_variance_ratio_
```

```python
varyans_oranlari
```

```python
print(f"1. Bileşen (PC1) Bilginin %{varyans_oranlari[0]*100:.2f}'ini temsil ediyor.")
print(f"2. Bileşen (PC2) Bilginin %{varyans_oranlari[1]*100:.2f}'ini temsil ediyor.")
print(f"Toplam Korunan Bilgi Oranı: %{sum(varyans_oranlari)*100:.2f}")
```
