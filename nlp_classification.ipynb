```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
```

```python
df=pd.read_csv("banka.csv")
df.head(5)
```

```python
mesaj=input("Lütfen yapmak istediğiniz işlemi giriniz.")
mesajdf=pd.DataFrame([{"metin":mesaj,"kategori":0}])
df=pd.concat([df,mesajdf,],ignore_index=True)
df["metin"]=df["metin"].str.lower()
df
```

```python
x=df["metin"].iloc[0:-1]
y=df["kategori"].iloc[0:-1]
tahmin=df["metin"].iloc[-1]
tahmin

```

```python
stop_words=['fakat','lakin','ancak','acaba', 'ama', 'aslında', 'az', 'bazı', 'belki', 'biri', 'birkaç', 'birşey', 'biz', 'bu', 'çok', 'çünkü', 'da', 'daha', 'de', 'defa', 'diye', 'eğer', 'en', 'gibi', 'hem', 'hep', 'hepsi', 'her', 'hiç', 'için', 'ile', 'ise', 'kez', 'ki', 'kim', 'mı', 'mu', 'mü', 'nasıl', 'ne', 'neden', 'nerde', 'nerede', 'nereye', 'niçin', 'niye', 'o', 'sanki', 'şey', 'siz', 'şu', 'tüm', 've', 'veya', 'ya', 'yani']
for word in stop_words:
    word=" "+word+" "
    df["metin"]=df["metin"].replace(word,"")
```

```python
cv=CountVectorizer(max_features=50)
x=cv.fit_transform(df["metin"].iloc[0:-1]).toarray()
tahmin=cv.transform([df["metin"].iloc[-1]]).toarray()
x
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,train_size=0.7)
rf=RandomForestClassifier()
model=rf.fit(x_train,y_train)
model.score(x_test,y_test)
```

```python
model.predict(tahmin)
```
