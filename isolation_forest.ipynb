```python
import pandas as pd
import numpy as np
import random
```

```python
df=pd.DataFrame()
```

```python
df["sensor"]=[4,3,4,4,3,11,2,3,4]
```

```python
deger=random.choice(range(df["sensor"].min(),df["sensor"].max()))
```

```python
deger
```

```python
alt=[]
ust=[]
for i in df["sensor"]:
    if i<deger:
        alt.append(i)
    else:
        ust.append(i)
```
